### Limpeza de ID_DOMICILIO

O codebook define `HHX` como identificador domiciliar, nao como medida numerica. O valor e mantido sem transformacao. Como um identificador ausente nao pode ser inferido com seguranca, a celula falha explicitamente se encontrar `NaN`.

```python

if ID_DOMICILIO not in df_limpo.columns:
    raise KeyError(f"Coluna selecionada ausente: {ID_DOMICILIO}")
if df_limpo[ID_DOMICILIO].isna().any():
    raise ValueError("ID_DOMICILIO tem valores ausentes; identificadores não são imputados.")
```

### Limpeza de PESO_AMOSTRAL

`WTFA_A` e o peso amostral final, numerico e positivo, sem codigos de recusa. Os valores sao mantidos; se houver ausencia pontual, ela e preenchida pela mediana observada, sem recodificar valores validos.

```python

df_limpo[PESO_AMOSTRAL] = pd.to_numeric(df_limpo[PESO_AMOSTRAL], errors="raise")
if df_limpo[PESO_AMOSTRAL].isna().any():
    mediana = df_limpo[PESO_AMOSTRAL].median()
    if pd.isna(mediana):
        raise ValueError("PESO_AMOSTRAL não tem valores observados para imputar.")
    df_limpo[PESO_AMOSTRAL] = df_limpo[PESO_AMOSTRAL].fillna(mediana)
assert np.isfinite(df_limpo[PESO_AMOSTRAL]).all() and (df_limpo[PESO_AMOSTRAL] > 0).all(), "PESO_AMOSTRAL deve ser finito e positivo."
```

### Limpeza de SEXO (SEX_A)

No codebook, `1` representa masculino e `2` feminino. Os códigos `7` (recusa), `8` (não apurado) e `9` (não sabe) são agrupados em `3` (resposta desconhecida). Valores ausentes (`NaN`) também recebem `3`. Assim, a variável fica sem valores ausentes e preserva as categorias válidas `1` e `2`.


```python

codigos_nao_resposta = {7: 3, 8: 3, 9: 3}
df_limpo[SEXO] = (
    df_limpo[SEXO]
    .replace(codigos_nao_resposta)
    .fillna(3)
    .astype(int)
)
assert df_limpo[SEXO].notna().all(), f"Ainda ha NaN em {SEXO}."
assert set(df_limpo[SEXO].unique()).issubset({1, 2, 3}), f"Codigo inesperado em {SEXO}."
```

### Limpeza de CLASSIFICACAO_URBANO_RURAL (URBRRL23)

O codebook apresenta quatro categorias validas (`1` a `4`), sem codigos especiais de nao resposta. Os codigos sao preservados como categorias; eventual `NaN` e imputado pela moda.

```python

serie = df_limpo[CLASSIFICACAO_URBANO_RURAL]
if serie.isna().any():
    moda = serie.mode(dropna=True)
    if moda.empty:
        raise ValueError(f"Não há resposta observada para imputar {CLASSIFICACAO_URBANO_RURAL!r}.")
    serie = serie.fillna(moda.iloc[0])
df_limpo[CLASSIFICACAO_URBANO_RURAL] = serie
assert set(df_limpo[CLASSIFICACAO_URBANO_RURAL].dropna().unique()).issubset({1, 2, 3, 4}), f"Código inesperado em {CLASSIFICACAO_URBANO_RURAL}."
assert df_limpo[CLASSIFICACAO_URBANO_RURAL].notna().all(), f"Ainda ha NaN em {CLASSIFICACAO_URBANO_RURAL}."
```

```python

if FREQ_SOLITARIO not in df_limpo.columns:
    if FREQ_SOLITARIO_CABECALHO_LEGADO in df_limpo.columns:
        df_limpo.rename(columns={FREQ_SOLITARIO_CABECALHO_LEGADO: FREQ_SOLITARIO}, inplace=True)
if FREQ_SOLITARIO not in df_limpo.columns:
    raise KeyError(f"Coluna selecionada ausente: {FREQ_SOLITARIO}")
df_limpo[FREQ_SOLITARIO] = df_limpo[FREQ_SOLITARIO].replace({1: 5, 2: 4, 3: 3, 4: 2, 5: 1, 7: 6, 8: 6, 9: 6}).fillna(6).astype(int)
assert df_limpo[FREQ_SOLITARIO].notna().all(), f"Ainda ha NaN em {FREQ_SOLITARIO}."
assert set(df_limpo[FREQ_SOLITARIO].unique()).issubset({1, 2, 3, 4, 5, 6}), f"Codigo inesperado em {FREQ_SOLITARIO}."
```
