### Avaliação de extremos com IQR

O IQR identifica valores distantes da distribuição, mas só é apropriado para medidas quantitativas contínuas. Neste conjunto, as medidas contínuas de peso corporal e altura foram removidas. As demais variáveis são categóricas, ordinais ou contagens agrupadas; por exemplo, o código `3` pode significar “3 ou mais”, e não exatamente três. Por isso, esses códigos não devem ser tratados como medidas contínuas nem avaliados com IQR.

A lista `colunas_quantitativas` fica vazia e, assim, o notebook não gera um relatório IQR para as variáveis atuais. A função é mantida para uso futuro, caso sejam incluídas medidas contínuas adequadas. Mesmo nesses casos, o IQR apenas sinaliza candidatos para revisão: não comprova erro nem remove ou altera valores. Para as variáveis atuais, a verificação de ruído deve se concentrar em códigos fora do codebook, frequências inesperadas e inconsistências entre respostas, preservando categorias raras que sejam válidas.


```python
def sinalizar_extremos_iqr(df, colunas, fator=1.5):
    resultados = []

    for coluna in colunas:
        valores = pd.to_numeric(df[coluna], errors="coerce").dropna()
        q1 = valores.quantile(0.25)
        q3 = valores.quantile(0.75)
        iqr = q3 - q1
        limite_inferior = q1 - fator * iqr
        limite_superior = q3 + fator * iqr
        extremos = valores[
            (valores < limite_inferior) | (valores > limite_superior)
        ]

        resultados.append({
            "variavel": coluna,
            "limite_inferior": limite_inferior,
            "limite_superior": limite_superior,
            "quantidade_sinalizada": len(extremos),
            "valores_sinalizados": extremos.unique().tolist(),
        })

    return pd.DataFrame(resultados)


if colunas_quantitativas:
    relatorio_extremos = sinalizar_extremos_iqr(
        df_silencioso,
        colunas_quantitativas,
    )
else:
    print("Sem variáveis quantitativas contínuas para análise por IQR.")
```
