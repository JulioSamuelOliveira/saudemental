### Correlação com a severidade da depressão (PHQ)

O teste de Pearson calcula o coeficiente de correlação e o p-valor para cada coluna numérica. Os p-valores são ajustados pelo método Benjamini-Hochberg (BH) para considerar os 71 testes simultâneos; use `p_ajustado_BH < 0,05` para identificar resultados significativos após o ajuste. A força da associação deve ser avaliada pelo coeficiente de correlação, não pelo p-valor.

```python
from pathlib import Path

import pandas as pd
from scipy.stats import false_discovery_control, pearsonr

arquivo_csv = Path("1-limpeza") / "dados" / "dados_limpos" / "adult25_limpo.csv"
if not arquivo_csv.exists():
    arquivo_csv = Path("..") / arquivo_csv
if not arquivo_csv.exists():
    raise FileNotFoundError(f"Não foi possível encontrar o arquivo: {arquivo_csv.resolve()}")

dados = pd.read_csv(arquivo_csv)
coluna_alvo = "SEVERIDADE_DEPRESSAO_PHQ"

if coluna_alvo not in dados.columns:
    raise KeyError(f"A coluna {coluna_alvo!r} não existe no arquivo selecionado.")

resultados = []
for coluna in dados.select_dtypes(include="number").columns:
    if coluna == coluna_alvo:
        continue

    amostra = dados[[coluna, coluna_alvo]].dropna()
    if len(amostra) < 2 or amostra[coluna].nunique() < 2 or amostra[coluna_alvo].nunique() < 2:
        correlacao, p_valor = float("nan"), float("nan")
    else:
        correlacao, p_valor = pearsonr(amostra[coluna], amostra[coluna_alvo])

    resultados.append(
        {
            "coluna": coluna,
            "correlacao_pearson": correlacao,
            "p_valor": p_valor,
            "n_amostras": len(amostra),
        }
    )

correlacoes = pd.DataFrame(resultados)
p_valores_validos = correlacoes["p_valor"].notna()
correlacoes.loc[p_valores_validos, "p_ajustado_BH"] = false_discovery_control(
    correlacoes.loc[p_valores_validos, "p_valor"].to_numpy(),
    method="bh",
)
correlacoes["significativo_BH_5%"] = correlacoes["p_ajustado_BH"] < 0.05
correlacoes = correlacoes.sort_values("p_ajustado_BH", na_position="last")

display(correlacoes)
```
