## Codificação one-hot das variáveis categóricas

As variáveis categóricas serão convertidas em colunas indicadoras binárias (one-hot encoding). A etapa incluirá variáveis nominais e ordinais codificadas em categorias; cada categoria observada terá sua própria coluna, portanto a ordem original das categorias ordinais não será preservada nessa representação. Códigos de resposta desconhecida ou não aplicável permanecerão como categorias distintas, sem serem confundidos com `Sim` ou `Não` e sem gerar valores nulos.

`ID_DOMICILIO` será mantido apenas como identificador para rastrear cada registro, e `PESO_AMOSTRAL` será preservado separadamente como metadado. Nenhum dos dois será incluído na matriz de atributos usada para calcular distâncias. A razão numérica de renda será mantida como variável numérica; sua categoria será codificada separadamente. Como ambas representam o mesmo conceito, essa sobreposição deverá ser considerada na interpretação dos clusters. A escala das variáveis numéricas será tratada em uma etapa posterior.

```python
colunas_metadados = [ID_DOMICILIO, PESO_AMOSTRAL]
colunas_numericas = [RAZAO_RENDA_POBREZA]
colunas_categoricas = [
    coluna
    for coluna in df_limpo.columns
    if coluna not in colunas_metadados + colunas_numericas
]

if df_limpo.isna().any().any():
    colunas_com_ausencias = df_limpo.columns[df_limpo.isna().any()].tolist()
    raise ValueError(f"A base ainda contém valores nulos antes do one-hot encoding: {colunas_com_ausencias}")
if not colunas_categoricas:
    raise ValueError("Nenhuma variável categórica foi identificada para codificação.")

df_codificado = pd.get_dummies(
    df_limpo,
    columns=colunas_categoricas,
    prefix=colunas_categoricas,
    prefix_sep="__",
    dtype=np.int8,
)

metadados_clusterizacao = df_codificado[colunas_metadados].copy()
X_clusterizacao = df_codificado.drop(columns=colunas_metadados).copy()

assert X_clusterizacao.notna().all().all(), "A matriz de atributos contém valores nulos."
assert all(pd.api.types.is_numeric_dtype(tipo) for tipo in X_clusterizacao.dtypes), "A matriz contém atributos não numéricos."
assert not set(colunas_categoricas).intersection(X_clusterizacao.columns), "Uma coluna categórica original permaneceu sem codificação."
assert len(X_clusterizacao) == len(metadados_clusterizacao) == len(df_limpo)

for coluna in colunas_categoricas:
    colunas_indicadoras = [nome for nome in df_codificado.columns if nome.startswith(f"{coluna}__")]
    if not colunas_indicadoras:
        raise ValueError(f"Nenhuma coluna one-hot foi criada para {coluna}.")
    if not df_codificado[colunas_indicadoras].sum(axis=1).eq(1).all():
        raise ValueError(f"Cada registro deve ter exatamente uma categoria ativa em {coluna}.")

print(f"Variáveis categóricas codificadas: {len(colunas_categoricas)}")
print(f"Atributos na matriz para clusterização: {X_clusterizacao.shape[1]}")
print(f"Metadados preservados separadamente: {', '.join(colunas_metadados)}")
print(f"Valores nulos na matriz: {int(X_clusterizacao.isna().sum().sum())}")
display(X_clusterizacao.head())
```

A matriz de atributos e os metadados serão exportados em arquivos separados, mantendo a correspondência de linhas. O arquivo de metadados permitirá relacionar os resultados dos clusters aos domicílios sem introduzir o identificador ou o peso amostral no cálculo das distâncias.

```python
pasta_saida_clusterizacao = Path("dados_clusterizacao")
pasta_saida_clusterizacao.mkdir(parents=True, exist_ok=True)

arquivo_atributos = pasta_saida_clusterizacao / "adult25_clusterizacao_onehot_2025.csv"
arquivo_metadados = pasta_saida_clusterizacao / "adult25_clusterizacao_metadados_2025.csv"
X_clusterizacao.to_csv(arquivo_atributos, index=False)
metadados_clusterizacao.to_csv(arquivo_metadados, index=False)

print(f"Matriz de atributos salva em: {arquivo_atributos}")
print(f"Metadados salvos em: {arquivo_metadados}")
```
